```python
#Import NumPy
import numpy as np
import time

print("NumPy version:", np.__version__)
```

```python
marks = np.array([
    [85, 78, 92],
    [70, 88, 76],
    [95, 91, 89],
    [65, 72, 80]
])

print("Marks Dataset:")
print(marks)
```

```python
print("First Student:")
print(marks[0])

print("\nFirst Student - First Subject:")
print(marks[0, 0])
```

```python
print("First Two Students:")
print(marks[:2])

print("\nFirst Subject Marks:")
print(marks[:, 0])
```

```python
print("Total Marks:", np.sum(marks))

print("\nMarks + 5:")
print(marks + 5)

print("\nMarks * 2:")
print(marks * 2)
```

```python
print("Average Marks:", np.mean(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))
print("Standard Deviation:", np.std(marks))
```

```python
print("Subject-wise Average:")
print(np.mean(marks, axis=0))

print("\nStudent-wise Average:")
print(np.mean(marks, axis=1))
```

```python
reshaped_marks = marks.reshape(2, 6)

print("Original Shape:", marks.shape)
print("New Shape:", reshaped_marks.shape)

print("\nReshaped Array:")
print(reshaped_marks)
```

```python
bonus = np.array([5, 3, 2])

result = marks + bonus

print("Original Marks:")
print(marks)

print("\nBonus:")
print(bonus)

print("\nAfter Broadcasting:")
print(result)
```

```python
np.save("marks.npy", marks)

print("Array saved successfully as marks.npy")
```

```python
loaded_marks = np.load("marks.npy")

print("Loaded Array:")
print(loaded_marks)
```

```python
numbers = list(range(1000000))
numpy_numbers = np.arange(1000000)

# Python List
start = time.time()

list_result = [x * 2 for x in numbers]

list_time = time.time() - start

# NumPy
start = time.time()

numpy_result = numpy_numbers * 2

numpy_time = time.time() - start

print("Python List Time:", list_time)
print("NumPy Time:", numpy_time)
```

```python
import matplotlib.pyplot as plt

students = ["Student 1", "Student 2", "Student 3", "Student 4", "Student 5"]
marks = [85, 72, 91, 68, 78]

plt.bar(students, marks)

plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```

```python
# Advanced NumPy Analysis

print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Maximum:", np.max(marks))
print("Minimum:", np.min(marks))
print("Standard Deviation:", np.std(marks))

print("Highest Marks Student Index:", np.argmax(marks))
print("Lowest Marks Student Index:", np.argmin(marks))
```

```python
import matplotlib.pyplot as plt

plt.hist(marks, bins=5)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
# Student Performance Analysis

students = ["Student 1", "Student 2", "Student 3", "Student 4", "Student 5"]

for student, mark in zip(students, marks):
    if mark >= 80:
        performance = "Excellent"
    elif mark >= 60:
        performance = "Good"
    elif mark >= 40:
        performance = "Pass"
    else:
        performance = "Fail"

    print(student, ":", mark, "-", performance)
```

```python
# Highest and Lowest Performing Student

highest_index = np.argmax(marks)
lowest_index = np.argmin(marks)

print("Highest Performing Student:")
print(students[highest_index], "-", marks[highest_index], "marks")

print("\nLowest Performing Student:")
print(students[lowest_index], "-", marks[lowest_index], "marks")
```

```python
# Find Highest and Lowest Marks

highest_index = np.argmax(marks)
lowest_index = np.argmin(marks)

print("Highest Performing Student:", students[highest_index])
print("Lowest Performing Student:", students[lowest_index])
```

```python
# Pass/Fail Summary

marks = np.array(marks)

pass_count = np.sum(marks >= 40)
fail_count = np.sum(marks < 40)

print("Total Students:", len(marks))
print("Passed Students:", pass_count)
print("Failed Students:", fail_count)
```

```python
labels = ["Pass", "Fail"]
counts = [pass_count, fail_count]

plt.bar(labels, counts)

plt.title("Pass/Fail Summary")
plt.xlabel("Result")
plt.ylabel("Number of Students")

plt.show()
```

```python
# Final Project Summary

print("===== NumPy Data Explorer =====")
print("Total Students:", len(marks))
print("Average Marks:", np.mean(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))
print("Passed Students:", pass_count)
print("Failed Students:", fail_count)

print("\nProject Analysis Completed Successfully!")
```
