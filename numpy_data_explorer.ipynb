```python
#Import NumPy
import numpy as np
import time

print("NumPy version:", np.__version__)
```

```python
marks = np.array([
    [85, 78, 92],
    [70, 88, 76],
    [95, 91, 89],
    [65, 72, 80]
])

print("Marks Dataset:")
print(marks)
```

```python
print("First Student:")
print(marks[0])

print("\nFirst Student - First Subject:")
print(marks[0, 0])
```

```python
print("First Two Students:")
print(marks[:2])

print("\nFirst Subject Marks:")
print(marks[:, 0])
```

```python
print("Total Marks:", np.sum(marks))

print("\nMarks + 5:")
print(marks + 5)

print("\nMarks * 2:")
print(marks * 2)
```

```python
print("Average Marks:", np.mean(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))
print("Standard Deviation:", np.std(marks))
```

```python
print("Subject-wise Average:")
print(np.mean(marks, axis=0))

print("\nStudent-wise Average:")
print(np.mean(marks, axis=1))
```

```python
reshaped_marks = marks.reshape(2, 6)

print("Original Shape:", marks.shape)
print("New Shape:", reshaped_marks.shape)

print("\nReshaped Array:")
print(reshaped_marks)
```

```python
bonus = np.array([5, 3, 2])

result = marks + bonus

print("Original Marks:")
print(marks)

print("\nBonus:")
print(bonus)

print("\nAfter Broadcasting:")
print(result)
```

```python
np.save("marks.npy", marks)

print("Array saved successfully as marks.npy")
```

```python
loaded_marks = np.load("marks.npy")

print("Loaded Array:")
print(loaded_marks)
```

```python
numbers = list(range(1000000))
numpy_numbers = np.arange(1000000)

# Python List
start = time.time()

list_result = [x * 2 for x in numbers]

list_time = time.time() - start

# NumPy
start = time.time()

numpy_result = numpy_numbers * 2

numpy_time = time.time() - start

print("Python List Time:", list_time)
print("NumPy Time:", numpy_time)
```
